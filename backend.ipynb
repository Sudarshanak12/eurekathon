```python
import pandas as pd
import numpy as np
```

```python
CATEGORIES = {
    "Individual": ["Food","Transport","Rent","Subscription","Entertainment","Investments"],
    "Business": ["Operational","Salaries","Utilities","Vendors","Tax","Inventory"],
    "Trip": ["Transport","Accommodation","Food","Local","Activities","Shopping"],
    "Family": ["Groceries","Fees","Healthcare","Insurance","Utilities","Maintenance"]
}
```

```python
def financial_score(df):
    total = df["Amount"].sum()
    savings = df[df["Amount"] < 0]["Amount"].sum()

    score = 100
    if total > 50000:
        score -= 25
    if savings > 0:
        score += 10
    return max(0,min(100,score))
```

```python
def detect_anomaly(df):
    mean = df["Amount"].mean()
    std = df["Amount"].std()

    df["Anomaly"] = df["Amount"].apply(
        lambda x: "⚠️ Unusual" if abs(x-mean) > 2*std else "Normal"
    )
    return df
```

```python
def summary(df):
    total = df["Amount"].sum()
    max_cat = df.groupby("Subcategory")["Amount"].sum().idxmax()

    return f"""
Total spending: ₹{total:.2f}
Highest spending category: {max_cat}

Advice:
Try reducing expenses in {max_cat}.
Track subscriptions and avoid unnecessary spending.
"""
```
